```python
def sum_1_to_n(n):
    res=0
    for i in range(1,n+1):          #summation of n --> n:5 ==> 1+2+3+4+5=15
        res+=i
    return res

sum_1_to_n(5)
```

```python
def lcm(n1,n2):
    lcm=None
    m=max(n1,n2)
    temp=m
    while True:
        if m%n1==0 and m%n2==0: #10--> 10,20,30,40,50,60--- #8-->8,16,24,32,40,48,56--
            lcm=m
            break
        m+=temp          # temp=1
    return lcm

lcm(10,8)
```

# 11. reverse 

```python
n=int(input())
temp=n
n=abs(n)
rev=0
while n > 0 :
    rem=n%10                 #fetching last digit                 #1234 --> 4321
    rev=rev*10+rem
    n//=10  #removing last digit
print(rev if temp>0 else -rev)
```

```python
def reverse(n):
    temp=n
    n=abs(n)
    rev=0
    while n > 0 :
        rem=n%10                 #fetching last digit                 #1234 --> 4321
        rev=rev*10+rem
        n//=10  #removing last digit              # return Str(n)[::-1] if n>0 else '-' +str(abs(n))[::-1]
    return rev if temp>0 else -rev

reverse(123)
```

# 12. palindrome

```python
n=int(input())
temp=n
rev=0
while n > 0 :
    rem=n%10                 #fetching last digit                 #def reverse(self,n):return Str(n)[::-1] if n>0 else '-' +str(abs(n))[::-1]
    rev=rev*10+rem                                                #def palindromic(self,n):
    n//=10  #removing last digit                                       #return self.reverse(n)==str(n)
if rev==temp:
    print(f'{temp} is a palindromic number')
else:
    print(f'{temp}is not a palindromic number')
```

# 13. count of zeros ,even,odd in n

```python
n = int(input())
zero_count = 0
even_count = 0
odd_count = 0
while n > 0:
    digit = n % 10   
    if digit == 0:
        zero_count += 1
    elif digit % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
    
    n = n // 10    

print(zero_count)
print(even_count)
print(odd_count)
```

```python
# n:153 --> power=3, cube(1)+cube(5)+cube(3)==1+125+27==153
n=int(input())
temp=n
sum=0
p=len(str(n))                  #temp=n sum=0 whilen>0: sum+=(n%10)**self.count_of_digits
while n >0:
    rem=n%10
    sum+=rem**p
    n//=10
if temp==sum:
    print(f'{temp} is an armstrong number')
else:
    print(f'{temp} is not an armstrong number')
```

#n is strong number or not

```python
n=int(input("n: "))
res=0
temp=n                 #145-->1!+4!+5!==145(strong)  #123-->1!+2!+3!==8(not strong)
while n>0:
    rem=n%10
    f=1
    for i in range(1,rem+1):
        f*=i
    res+=f
    n//=10
print("strong num" if temp==res else "not a strong num")
```
