# 1. count of vowels and consonents 

```python
s="Python coding is Awesome"
o=0
c=0
for i in s:
    if i==" ":
        continue
    elif i in 'aeiouAEIOU':
        o+=1
    else:
        c+=1
print("vowels count",o)
print("consonents count",c)
```

# 2.upper to lower.lower to upper

```python
s="PyCoDing Is AweSoMe"
print(s)
p=""
for i in s :
    if i==' ':
       p+=i
    elif ord(i) >=65 and ord(i) <=90:
       p+= chr(ord(i)+32)
    elif ord(i) >=97 and ord(i) <=122:
       p+= chr(ord(i)-32)

print(p)
```

```python
s1= "friend_in_need_is_a_friend_indeed"
print(s1)
s2=""
s2+=chr(ord(s1[0])-32)
for i in range(1,len(s1)):
    if s1[i]=='_':
        s2+=" "
    elif s1[i-1]=="_":
        s2+=chr(ord(s1[i])-32)
    else:
        s2+= s1[i]
print(s2)
```

wap to check the given strings are anagram strings are not
s1="silent"
s2="listen" (same letters in both strings)

```python

def bubble_sort(s):
    l1=list(s)
    for i in range(len(l1)):
        for j in range(i+1,len(l1)):
            if l1[i] > l1[j]:
                l1[i],l1[j]=l1[j],l1[i]
    return "".join(l1)
s1="silent"
s2="listen"
if bubble_sort(s1)==bubble_sort(s2):
    print("anagram strings")
else:
    print("not anagram strings")
```

panagram (all 26 letters of alphabtes in a given sentence)

```python
s1='The quick brown fox jumps over a lazy Dog'
#s1='pack my box with Five Dozen Liquor Jugs'
res=set()
for i in s1:
    if i ==" ":
        continue
    elif i not in res:
        res.add(i.upper())
if len(res) ==26:
    print("panagram")
else:
    print("not panagram")
```
