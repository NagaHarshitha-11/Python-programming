program to rotate the list to left side for n number of times(list left rotation)

```python
# 11,22,33,44,55 ---> 22,33,44,55,11 (rotation for 1st time on left side)--> 33,44,55,11,22(rotation 2 times on left side)
l=[11,22,33,44,55]
n=int(input("enter rotation: "))
for _ in range(n):
    temp=l[0]
    for i in range(len(l)-1):
        l[i]=l[i+1]
    l[len(l)-1]=temp
print(l)
```

bubble sort(sinking sort)
take 2 variables i and j
if i is pointing to first index , j should always points to its next index
compare i and j values , if i th value is greater than j th value , swap i th and j th value

```python
def bubble_sort(l):
    for i in range(len(l)):
        for j in range(i+1,len(l)):
            if l[i]>l[j]:
                l[i],l[j]=l[j],l[i]
l=[3,5,1,6,7,2]
print("bubble sort")
print("before sorting",l)
bubble_sort(l)
print("after sorting",l)
```

wap to find sum of all prime numbers in the given list using functions

```python
def is_prime(val):
    if val==1:
        return False
    else:
        for i in range(2,val):
            if val%i==0:
                return False
        return True
l1=[1,2,11,8,6,9,15,17,2,13,21]
cnt=0
for i in l1:
    if is_prime(i):
        cnt+=i
print("sum of all prime numbers",cnt)
```
